```python
from ultralytics import YOLO
```

```python
DATA_DIR = "dataset"         
IMG_SIZE = 224
BATCH_SIZE = 16                
EPOCHS = 30
MODEL_SIZE = "yolov8n-cls.pt"  
```

```python
model = YOLO(MODEL_SIZE)  

results = model.train(
    data=DATA_DIR,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH_SIZE,
    device="0",          
    patience=8,             
    project="waste_classifier_runs",
    name="yolov8_waste_cls",
    exist_ok=True,
    augment=True,           
    val=True,
    plots=True,             
)

print("\nTraining complete.")
print("Best model saved at: waste_classifier_runs/yolov8_waste_cls/weights/best.pt")
```

```python
best_model_path = ""

model = YOLO(best_model_path)
metrics = model.val(data=DATA_DIR)
print(metrics)
```

```python
from ultralytics import YOLO

# Load your trained model
model = YOLO(r"C:\Users\KISHOR SELVA\Desktop\Outlier\backend\model\waste_classifier.pt")

# Test on one image
results = model.predict(
    source=r"C:\Users\KISHOR SELVA\Desktop\Outlier\test_image\test15.jpg",
    imgsz=320,
    verbose=False
)

result = results[0]

# Get prediction
class_id = result.probs.top1
class_name = result.names[class_id]
confidence = float(result.probs.top1conf)

print("Predicted Class :", class_name)
print(f"Confidence      : {confidence*100:.2f}%")

print("\nAll Class Probabilities:")
for i, prob in enumerate(result.probs.data.tolist()):
    print(f"{result.names[i]} : {prob*100:.2f}%")
```
